```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
import seaborn as sns

# Load the dataset directly as a Pandas DataFrame
df = sns.load_dataset('iris')
df
```

```python
df.info()
df.describe()
df.isnull().sum()

```

```python
plt.figure(figsize=(10,6)) 
plt.scatter(df['sepal_length'], df['sepal_width'])
plt.legend(df['species'])
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
plt.title('Relationship between Sepal Length and Sepal Width')
plt.show()
```

```python
plt.figure(figsize=(10,6)) 
plt.scatter(df['petal_length'], df['sepal_width'])
plt.legend(df['species'])
plt.xlabel('Petal Length')
plt.ylabel('Sepal Width')
plt.title('Relationship between Petal Length and Sepal Width')
plt.show()
```

```python
plt.figure(figsize=(10,6)) 
plt.scatter(df['petal_length'], df['petal_width'])
plt.legend(df['species'].unique())
plt.xlabel('Petal Length')
plt.ylabel('Petal Width')
plt.title('Relationship between Petal Length and Petal Width')
plt.show()
```
