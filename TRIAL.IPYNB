```python
plt.figure(figsize=(10,6))
plt.hist(df['Profit'], bins=20, color='skyblue', edgecolor='black')
plt.xlabel('Profit')
plt.ylabel('Frequency')
plt.title('Distribution of Profit')
plt.show()
```

```python
#city with the highest profit
plt.figure(figsize=(10,6))
city_profit = df.groupby('City')['Profit'].sum().sort_values(ascending=False)
sns.barplot(x=city_profit.index, y=city_profit.values, palette='viridis')
plt.xticks(rotation=45)
plt.xlabel('City')
plt.ylabel('Total Profit')
plt.title('Total Profit by City')
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.scatter(df['Profit'], df['Sales'])
plt.xlabel('Profit')
plt.ylabel('Sales')
plt.title('Relationship between Profit and Sales')
plt.show()
```
